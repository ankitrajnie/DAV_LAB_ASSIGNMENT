# Introduction to Python for Data Analytics

Python provides powerful libraries for data analytics. Two commonly used libraries are **NumPy** and **Pandas**.

- **NumPy** → Numerical calculations and arrays
- **Pandas** → Data manipulation and analysis

## 1. NumPy Basics

NumPy (Numerical Python) is used to work with arrays and perform mathematical operations efficiently.

```python
import numpy as np
```

### Creating a NumPy Array

```python
arr = np.array([10, 20, 30, 40, 50])
print(arr)
print(type(arr))
```

## 2. 2D NumPy Array

A 2D array contains rows and columns.

```python
matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(matrix)
print("Shape:", matrix.shape)
```

## 3. NumPy Array Operations

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```

## 4. Useful NumPy Functions

```python
arr = np.array([10, 20, 30, 40, 50])

print("Sum:", np.sum(arr))
print("Mean:", np.mean(arr))
print("Maximum:", np.max(arr))
print("Minimum:", np.min(arr))
```

```python
print("Zeros:", np.zeros(5))
print("Ones:", np.ones(5))
print("Range:", np.arange(1, 10))
```

## 5. NumPy Array Indexing

Indexing is used to access individual elements.

```python
arr = np.array([10, 20, 30, 40, 50])

print(arr[0])
print(arr[2])
print(arr[-1])
```

## 6. NumPy Array Slicing

Slicing is used to select a range of elements.

```python
arr = np.array([10, 20, 30, 40, 50])

print(arr[1:4])
print(arr[:3])
print(arr[2:])
```

# 7. Pandas Basics

Pandas is a Python library used for working with structured and tabular data.

```python
import pandas as pd
```

## 8. Creating a Pandas Series

A Series is a one-dimensional data structure.

```python
marks = pd.Series([85, 90, 78, 92])
print(marks)
```

## 9. Creating a DataFrame

A DataFrame is a two-dimensional table containing rows and columns.

```python
data = {
    "Name": ["Adarsh", "Rahul", "Ankit"],
    "Age": [20, 21, 19],
    "Marks": [85, 90, 78]
}

df = pd.DataFrame(data)
print(df)
```

## 10. Basic DataFrame Information

```python
print(df.head())
print("Shape:", df.shape)
print("Columns:", df.columns)
```

## 11. DataFrame Indexing

### Selecting a Single Column

```python
print(df["Name"])
```

### Selecting Multiple Columns

```python
print(df[["Name", "Marks"]])
```

## 12. Using `loc`

`loc` is used for label-based indexing.

```python
print(df.loc[0])
print(df.loc[0, "Name"])
```

## 13. Using `iloc`

`iloc` is used for position-based indexing.

```python
print(df.iloc[0])
print(df.iloc[0, 1])
```

## 14. DataFrame Slicing

```python
# Select the first two rows
print(df[0:2])

# Select the first two rows and first two columns
print(df.iloc[0:2, 0:2])
```

## 15. Filtering Data

Pandas can filter data based on conditions.

```python
high_marks = df[df["Marks"] > 80]
print(high_marks)
```
