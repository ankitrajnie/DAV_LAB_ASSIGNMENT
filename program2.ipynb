# Exploratory Data Analysis (EDA)

Exploratory Data Analysis (EDA) is the process of examining a dataset to understand its structure, summarize important information, and identify problems such as missing values and duplicate records.

### Topics Covered
- Loading datasets
- Computing descriptive statistics
- Identifying missing values
- Identifying duplicate records
- Basic data cleaning

## 1. Import Pandas

```python
import pandas as pd
```

## 2. Create a Hypothetical Dataset

To demonstrate EDA, we will create a small hypothetical student dataset.

The dataset intentionally contains:
- One missing `Age`
- One missing `Marks`
- One duplicate row

This allows us to practice identifying common data-quality problems.

```python
data = {
    "Name": ["Adarsh", "Rahul", "Ankit", "Priya", "Sneha", "Rahul"],
    "Age": [20, 21, 19, 20, None, 21],
    "Marks": [85, 90, 78, None, 88, 90],
    "City": ["Mysuru", "Bengaluru", "Mysuru", "Delhi", "Bengaluru", "Bengaluru"]
}

df = pd.DataFrame(data)

print(df)
```

In a real data-analysis project, data is often stored in a CSV file.

## 4. Understanding the Dataset

`head()` displays the first few rows, `shape` gives the number of rows and columns, and `columns` gives the column names.

```python
print("First 5 rows:")
print(df.head())

print("\nDataset shape:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())
```

## 5. Dataset Information

`info()` displays the number of entries, columns, non-null values, and data types.

```python
df.info()
```

## 6. Descriptive Statistics

`describe()` provides statistical information for numerical columns.

```python
print(df.describe())
```

### Mean, Median and Mode

These are common descriptive statistics:
- **Mean** → Average value
- **Median** → Middle value
- **Mode** → Most frequently occurring value

```python
print("Mean Marks:", df["Marks"].mean())
print("Median Marks:", df["Marks"].median())
print("Mode Marks:", df["Marks"].mode()[0])
```

## 7. Identifying Missing Values

`isnull()` returns `True` wherever a value is missing.

```python
print(df.isnull())
```

### Count Missing Values

`isnull().sum()` counts missing values in each column.

```python
print(df.isnull().sum())
```

### Display Rows Containing Missing Values

```python
print(df[df.isnull().any(axis=1)])
```

## 8. Identifying Duplicate Records

`duplicated()` identifies rows that are exact duplicates of an earlier row.

```python
print(df.duplicated())
```

### Count Duplicate Records

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

### Display Duplicate Rows

```python
print(df[df.duplicated()])
```

## 9. Removing Duplicate Records

`drop_duplicates()` removes duplicate rows from the DataFrame.

```python
df_clean = df.drop_duplicates()

print(df_clean)
print("\nRows after removing duplicates:", len(df_clean))
```

## 10. Handling Missing Values

Missing values can be replaced using `fillna()`.

Here, missing `Age` and `Marks` values are replaced with the respective column means.

```python
df_filled = df_clean.copy()

df_filled["Age"] = df_filled["Age"].fillna(df_filled["Age"].mean())
df_filled["Marks"] = df_filled["Marks"].fillna(df_filled["Marks"].mean())

print(df_filled)
```

## 11. Final EDA Check

After cleaning, we can check whether any missing values or duplicate rows remain.

```python
print("Missing values:")
print(df_filled.isnull().sum())

print("\nDuplicate rows:", df_filled.duplicated().sum())
```

# Quick Summary

| Function | Purpose |
|---|---|
| `pd.read_csv()` | Loads a CSV dataset |
| `head()` | Displays the first rows |
| `shape` | Returns rows and columns |
| `info()` | Shows dataset information |
| `describe()` | Generates descriptive statistics |
| `isnull()` | Identifies missing values |
| `isnull().sum()` | Counts missing values |
| `duplicated()` | Identifies duplicate rows |
| `drop_duplicates()` | Removes duplicate rows |
| `fillna()` | Replaces missing values |
| `dropna()` | Removes rows containing missing values |

### EDA Workflow

**Load Data → Understand Data → Calculate Statistics → Find Missing Values → Find Duplicates → Clean Data → Verify Data**
