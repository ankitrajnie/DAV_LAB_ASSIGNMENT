# Program 3: Data Visualization using Matplotlib and Seaborn

This program demonstrates histograms, box plots, scatter plots, pair plots, and correlation heatmaps using a student performance dataset.

**Dataset:** `student_performance_dataset.csv`



```python
# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


# ============================================================
# 2. LOAD STUDENT DATASET
# ============================================================

df = pd.read_csv("student_performance_dataset.csv")


# ============================================================
# 3. DISPLAY DATA
# ============================================================

print("Student Performance Dataset:")
display(df)


# ============================================================
# 4. HISTOGRAM
# ============================================================

plt.figure(figsize=(7, 4))
sns.histplot(df["Marks"], bins=6, kde=True)
plt.title("Histogram of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.show()


# ============================================================
# 5. BOX PLOT
# ============================================================

df["Study_Level"] = pd.cut(
    df["Study_Hours"],
    bins=[0, 3, 5, 10],
    labels=["Low", "Medium", "High"]
)

plt.figure(figsize=(7, 4))
sns.boxplot(x="Study_Level", y="Marks", data=df)
plt.title("Box Plot of Marks by Study Level")
plt.xlabel("Study Level")
plt.ylabel("Marks")
plt.show()


# ============================================================
# 6. SCATTER PLOT
# ============================================================

plt.figure(figsize=(7, 4))
sns.scatterplot(
    x="Study_Hours",
    y="Marks",
    data=df,
    s=70
)
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()


# ============================================================
# 7. PAIR PLOT
# ============================================================

sns.pairplot(
    df[["Study_Hours", "Attendance", "Assignment_Score", "Marks"]]
)
plt.suptitle(
    "Pair Plot of Student Performance Variables",
    y=1.02
)
plt.show()


# ============================================================
# 8. CORRELATION HEATMAP
# ============================================================

correlation = df[
    ["Study_Hours", "Attendance", "Assignment_Score", "Marks"]
].corr()

plt.figure(figsize=(7, 5))
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Correlation Heatmap")
plt.show()


# ============================================================
# 9. CONCLUSION
# ============================================================

print("Visualization analysis completed successfully.")
```

## Key Observations

- The histogram shows the distribution of student marks.
- The box plot compares marks across different study-hour levels.
- The scatter plot shows the relationship between study hours and marks.
- The pair plot compares relationships among the numerical performance variables.
- The correlation heatmap shows the strength and direction of relationships between the numerical variables.

## Conclusion

Matplotlib and Seaborn make it easier to understand distributions, relationships, patterns, and correlations in student performance data.
